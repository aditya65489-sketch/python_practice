# MATPLOTLIB Tutorial


```python
from matplotlib import pylab
print(pylab.__version__)
```

```python
"""Use numPy to genrate random data"""
import numpy as np

x = np.linspace(0, 10,25)
y = x * x +2
print(x)
print(y)
print(np.array([x, y]).reshape(25,2))
```

```python
"""It only take 1 comand to draw """
pylab.plot(x,y, 'r') #r stands for red
```

```python
"""Drawing a subgraph"""
pylab.subplot(1,2,1) #The contents of the brackets represents (rows, columns, indexs)
pylab.plot(x,y, 'r--') #the thrid parameter here determines color and line style
pylab.subplot(1,2,2)
pylab.plot(y,x, 'g*-')
```

```python
#(1,2,1) and (1,2,2) are significant if we want tilled subplot , without which both the graph will in the same plot

pylab.subplot(1,2,1)# The contents of brackets represents (rows, columns, indexs0
pylab.plot(x,y, 'r--')# The third parameter here determines color and line stle
pylab.subplot(1,2, 1)
pylab.plot(y,x, 'g*-')
```
