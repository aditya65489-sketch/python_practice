Nested if statement

```python
c=50
if c<25:
    if c%2==0:
        print("c is an even number less than 25")
    else:
        print("c is an odd number less than 25")
else:
    print("c is grater than 25")
         
```

if-elif-else ladder

```python
var="z"
if var=='a':
    print("This is vowel a")
elif var=='e':
    print("This is vowel e")
elif var=='i':
    print("This is vowel i")
elif var=='o':
    print("This is vowel o")
elif var=='u':
    print("This is vowel u")
else:
    print("This is consonent")
```
