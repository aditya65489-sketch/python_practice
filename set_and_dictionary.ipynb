Sets

```python
s=set([1,2,3,4])
print(s)
print(type(s))
```

```python
s.add('a')
```

```python
print(s)
```

```python
fs=frozenset([1,2,3,4])
print(fs)
```

```python
fs.add("a")
```

```python
s1=set([1,3,5,2])
s2=set([3,2,8,6])
```

```python
s1.union(s2)
```

```python
s1.difference(s2)
```
